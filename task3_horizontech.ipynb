## 4. Salary Analysis

Salary is one of the important variables in the dataset.
The standardized salary in USD is analyzed to understand its
distribution and relationship with different job characteristics.

```python
#Salary Statistics
print("Salary Statistics:")
display(df['salary_in_usd'].describe())
```

**Salary Distribution**

```python

plt.figure(figsize=(8, 5))

sns.histplot(df['salary_in_usd'], bins=30, kde=True)

plt.title('Salary Distribution in USD')
plt.xlabel('Salary (USD)')
plt.ylabel('Frequency')

plt.show()
```

**Salary Outliers**

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df['salary_in_usd'])

plt.title('Salary Outlier Analysis')
plt.xlabel('Salary (USD)')

plt.show()
```

## 5. Experience Level Analysis

This section examines how experience level is distributed across
technology jobs and how experience is associated with salary.

**Experience Distribution**

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='experience_level')

plt.title('Job Distribution by Experience Level')
plt.xlabel('Experience Level')
plt.ylabel('Number of Jobs')

plt.show()
```

**Average Salary by Experience**

```python
experience_salary = (
    df.groupby('experience_level')['salary_in_usd']
    .agg(['count', 'mean', 'median'])
    .sort_values('mean', ascending=False)
)

experience_salary.columns = [
    'Number of Jobs',
    'Average Salary',
    'Median Salary'
]

display(experience_salary)
```

**Experience vs Salary**


```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x='experience_level',
    y='salary_in_usd'
)

plt.title('Salary Distribution by Experience Level')
plt.xlabel('Experience Level')
plt.ylabel('Salary (USD)')

plt.show()
```

## 6. Job Role Analysis

This section analyzes the distribution of technology job roles in the dataset.
The analysis identifies the most common job titles and compares their
average salaries.

```python
top_jobs = df['job_title'].value_counts().head(15)

print("Top 15 Most Common Job Roles:")
display(top_jobs)
```

**Top Job Roles Visualization**

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_jobs.values,
    y=top_jobs.index
)

plt.title('Top 15 Most Common Technology Job Roles')
plt.xlabel('Number of Jobs')
plt.ylabel('Job Title')

plt.show()
```

**Average Salary by Job Role**

```python
job_salary = (
    df.groupby('job_title')['salary_in_usd']
    .agg(['count', 'mean', 'median'])
)

job_salary.columns = [
    'Number of Jobs',
    'Average Salary',
    'Median Salary'
]

job_salary = job_salary.sort_values(
    'Average Salary',
    ascending=False
)

display(job_salary.head(15))
```

**Highest Paying Job Roles**

```python
highest_paying_jobs = (
    df.groupby('job_title')['salary_in_usd']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 Highest-Paying Job Roles:")
display(highest_paying_jobs)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=highest_paying_jobs.values,
    y=highest_paying_jobs.index
)

plt.title('Top 10 Highest-Paying Technology Job Roles')
plt.xlabel('Average Salary (USD)')
plt.ylabel('Job Title')

plt.show()
```

## 7. Work Model Analysis

This section examines the distribution of remote, hybrid, and on-site
work models and compares salary levels across different work arrangements.

**Work Model Distribution**

**Salary by Work Model**

## 8. Company Size Analysis

This section examines the distribution of technology jobs across small,
medium, and large companies. It also compares salary levels across
different company sizes.

**Company Size Distribution**

```python
company_size_counts = df['company_size'].value_counts()

print("Company Size Distribution:")
display(company_size_counts)
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='company_size'
)

plt.title('Distribution of Jobs by Company Size')
plt.xlabel('Company Size')
plt.ylabel('Number of Jobs')

plt.show()
```

**Salary by Company Size**

**Salary Visualization**

## 9. Geographic Analysis

This section analyzes technology job locations to identify countries
with a higher number of job opportunities and compare salary levels
across different locations.

**Top Job Locations**

```python
top_locations = df['company_location'].value_counts().head(15)

print("Top 15 Company Locations by Number of Jobs:")
display(top_locations)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_locations.values,
    y=top_locations.index
)

plt.title('Top 15 Company Locations by Number of Jobs')
plt.xlabel('Number of Jobs')
plt.ylabel('Company Location')

plt.show()
```

**Average Salary by Location**

```python
location_salary = (
    df.groupby('company_location')['salary_in_usd']
    .agg(['count', 'mean', 'median'])
)

location_salary.columns = [
    'Number of Jobs',
    'Average Salary',
    'Median Salary'
]

# Consider locations with at least 20 jobs
location_salary_filtered = (
    location_salary[location_salary['Number of Jobs'] >= 20]
    .sort_values('Average Salary', ascending=False)
)

display(location_salary_filtered.head(15))
```

**Highest Paying Locations**

```python
top_paying_locations = (
    location_salary_filtered['Average Salary']
    .head(10)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_paying_locations.values,
    y=top_paying_locations.index
)

plt.title('Top 10 Highest-Paying Company Locations')
plt.xlabel('Average Salary (USD)')
plt.ylabel('Company Location')

plt.show()
```

## 10. Year-wise Salary and Job Market Trends

This section analyzes changes in the technology job market across
different years. The number of jobs and salary statistics are compared
to identify overall trends.

**Year Analysis**

```python
year_analysis = (
    df.groupby('work_year')['salary_in_usd']
    .agg(['count', 'mean', 'median'])
)

year_analysis.columns = [
    'Number of Jobs',
    'Average Salary',
    'Median Salary'
]

display(year_analysis)
```

**Number of Jobs by Year**

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='work_year'
)

plt.title('Number of Technology Jobs by Year')
plt.xlabel('Year')
plt.ylabel('Number of Jobs')

plt.show()
```

**Average Salary Trend**

```python
plt.figure(figsize=(8, 5))

sns.lineplot(
    data=year_analysis,
    x=year_analysis.index,
    y='Average Salary',
    marker='o'
)

plt.title('Average Technology Salary Trend')
plt.xlabel('Year')
plt.ylabel('Average Salary (USD)')

plt.show()
```

**Median Salary Trend**

```python
plt.figure(figsize=(8, 5))

sns.lineplot(
    data=year_analysis,
    x=year_analysis.index,
    y='Median Salary',
    marker='o'
)

plt.title('Median Technology Salary Trend')
plt.xlabel('Year')
plt.ylabel('Median Salary (USD)')

plt.show()
```

## 12. Job Demand vs Salary Analysis

This analysis compares the demand for different technology job roles
with their average salaries. Only job roles with sufficient records
are considered to avoid misleading conclusions from very small samples.

**Create Analysis Table**

```python
job_demand_salary = (
    df.groupby('job_title')['salary_in_usd']
    .agg(['count', 'mean'])
)

job_demand_salary.columns = [
    'Job Demand',
    'Average Salary'
]

# Consider job roles with at least 20 records
job_demand_salary = job_demand_salary[
    job_demand_salary['Job Demand'] >= 20
]

display(
    job_demand_salary.sort_values(
        'Job Demand',
        ascending=False
    ).head(15)
)
```

```python
top_demand_jobs = (
    job_demand_salary
    .sort_values('Job Demand', ascending=False)
    .head(15)
)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=top_demand_jobs,
    x='Job Demand',
    y='Average Salary',
    s=100
)

plt.title('Job Demand vs Average Salary')
plt.xlabel('Number of Job Records')
plt.ylabel('Average Salary (USD)')

plt.show()
```

**Highest Demand Jobs**

```python
print("Highest Demand Job Roles:")

display(
    job_demand_salary
    .sort_values('Job Demand', ascending=False)
    .head(10)
)
```

## 13. Correlation Analysis

Correlation analysis is used to examine the relationships between
numerical variables in the dataset. This helps identify whether
variables show positive or negative relationships.

```python
#Correlation Matrix
numeric_columns = df.select_dtypes(
    include=np.number
)

correlation_matrix = numeric_columns.corr()

display(correlation_matrix)
```

**Correlation Heatmap**

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Numerical Variables')

plt.show()
```
